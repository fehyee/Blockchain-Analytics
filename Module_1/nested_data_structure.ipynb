### Nested Data Structure

These are combinations of basic data structures. They are common in real world data (like blockchain data, API responses, and JSON)

### List of Lists (2D Lists)
Can be used in tabular data, eg. Spreadsheets, CSV  

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print(matrix[2][0])
print(matrix[2][2])
print(matrix[0][2])
```

### List of Dictionaries

```python
user = [
    {'name' : 'Alice', 'wallet' : '0xabc123'},
    {'name' : 'Bob', 'wallet' : '0xdef456'},
    {'name' : 'Charlie', 'wallet' : '0xghi789'}
]

print(user[1]['wallet'])  # Accessing Bob's wallet
print(user[2]['name'])  # Accessing Charlie's name
```

### Dictionary of Lists

```python
wallets = {
    '0xabc123' : ['ETH', 'BTC', 'SOL'],
    '0xdef456' : ['ETH', 'LTC'],
    '0xghi789' : ['ETH', 'XRP', 'ADA']
}

print(wallets['0xdef456'])  # Accessing the list of coins in Bob's wallet
print(wallets['0xabc123'][2]) # Accessing the third coin in Alice's wallet
```

### Dictionary of Dictionary

```python
users_data = {
    '0xabc123' : {'name': 'Alice', 'balance': 30},
    '0xdef456' : {'name': 'Bob', 'balance': 20},
    '0xghi789' : {'name': 'Charlie', 'balance': 50}
}

print(users_data['0xabc123']['balance'])
```
