### Sequence Data Type
1. List - Created using []. It is an ordered collection of items.
2. Tuple - Created using (). It is an ordered collection of items, but imutable.
3. dictionary - created using {}. It only stores 'keys' and 'values'. A collection of key-value pair
4. Set - Also created using {}. An unordered collection of unique items. No duplicates


```python
name = "Alice"
age = 30
height = 5.5
list = [1, 2, 3, 4, 5]
tuple = (1, 2, 3, 4, 5)
dictionary = {"name": "Alice", "age": 30, "crypto": "Bitcoin"}
```

### Checking Data Type

```python
print("Data type:", type(name))
print("Data type:", type(age))  
print("Data type:", type(height))
print("Data type:", type(list))  
print("Data type:", type(tuple))
print("Data type:", type(dictionary))
```

### Converting Data Type


```python
age_str = "30"
age_int = int(age_str)
print("Converted age:", age_int)
age_float = float(age_str)
print("Converted age to float:", age_float)
```

### Lists
### Common Methods


```python

tokens = ["ETH", "SOL", "BTC"]
tokens 
```

```python
# len() -> returns the length of a string
len(tokens)
```

```python
# .append() -> adds an element to the end of a list
tokens.append("DAI")
print(tokens)
```

```python
# .insert(index, value) -> inserts an element at a specific index
tokens.insert(5, "XRP")
print(tokens)
```

```python
# .remove() -> removes an the first match of an element from a list
tokens.remove("XRP")
print(tokens)
```

```python
# .pop() -> removes and return the last element from a list
# Or # tokens.pop(0) -> removes and returns the first element from a list
# tokens.pop(2) -> removes and returns the third element from a list

tokens.pop() 
print(tokens)   
```

```python
# .sort() -> sorts a list in ascending order
tokens.sort()
print(tokens)
```

```python
# .reverse() -> reverses the order of a list
tokens.reverse()
print(tokens)
```

```python
tokens.count("ETH")  # Counts the occurrences of "ETH" in the list
```

### Indexing and Slicing

```python
# Indexing

T = ["ETH", "SOL", "BTC", "DAI", "XRP"]
T[0]  # First element
```

```python
# Slicing

# T[1:3]  # Elements from index 1 to 2 (not including 3)
# T[:3]  # First three elements
# T[2:]  # Elements from index 2 to the end
# T[-1]  # Last element
# T[-2:]  # Last two elements
T[::2]  # Every second element
# T[::-1]  # Reversed list
# T[1:4:2]  # Elements from index 1 to 3, stepping by 2
```

### Dictionary
The key must be unique

```python
# Key-Value Pairs
wallet = {
    "ETH": 2.5,
    "SOL": 10.0,
    "BTC": 0.1,
    "DAI": 500.0
}
```

```python
# dict{key} -> Access value by key

# print(Wallet["ETH"])  # Access value by key
wallet['ETH']
```

```python
# .get(key) -> Access value by key, returns None if key not found
wallet.get('ETH')  # Access value by key, returns None if key not found
```

```python
# .keys() -> returns a list of keys in the dictionary
wallet.keys()  # Returns a list of keys in the dictionary
```

```python
# .values() -> returns a list of values in the dictionary
wallet.values()  # Returns a list of values in the dictionary
```

```python
# .items() -> returns a list of key-value pairs in the dictionary
wallet.items()  # Returns a list of key-value pairs in the dictionary
```

```python
# .update(key, value) -> updates the value for a key, or adds the key-value pair if it doesn't exist
wallet.update({"ADA": 1000.0})  # Adds a new key-value pair
wallet 
```

```python
# .pop() -> removes the last element from a list
# .pop(key) -> removes the key-value pair for the specified key

wallet.pop("ADA")  # Removes the key-value pair for "ADA"
wallet
```

### Sets


```python
Address = {'0xabc123', '0xdef456', '0xghi789'}
```

```python
# .add() -> adds an element to a set

Address.add('0xjkl012')  # Adds a new address to the set
Address
```

```python
# .remove() -> removes an element from a set
Address.remove('0xjkl012')  # Removes the address '0xjkl012' from the set
Address
```

```python
# .discard() -> removes an element from a set, does not raise an error if the element is not found
Address.discard('0xabc123')  # Removes the address '0xabc123' from the set, does not raise an error if not found
Address
```

```python
# .union() -> returns a new set with elements from both sets
Address2 = {'0xxyz123', '0xabc123', '0xdef456'}
Address.union(Address2)  # Returns a new set with elements from both sets
```

```python
# .intersection() -> returns a new set with elements common to both sets
Address.intersection(Address2)  # Returns a new set with elements common to both sets
```

```python
# .difference() -> returns a new set with elements in the first set but not in the second set

Address.difference(Address2)  # Returns a new set with elements in Address but not in Address2
```

```python
'0xghi789' in Address  # Checks if '0xghi789' is in the set Address
```
