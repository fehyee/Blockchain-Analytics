## Working with files
-  Reading & Writing Text Files

### Basic Operations  
- open("file.txt") - Opens a file (needs to be closed after)  
- with open("file.txt") - Better way to auto close the file

### Modes
- "r" - Read (default)
- "w" - Write (erases exiting content)
- "a" - append (adds to end)
- "r+" - Read & Write

```python
# Reading Mode

with open('Sample.txt', 'r') as file: # Opens the file in read mode
    # Reads the content of the file
    content = file.read()
    print(content)
```

```python
# Writing Mode
with open('Sample.txt', 'w') as file: # Opens the file in write mode
    # Writes new content to the file
    file.write("This is a new line of text.\n")
    print("Content written to file.")
```

```python
# Append mode
with open('Sample.txt', 'a') as file: # Opens the file in append mode
    # Appends new content to the file
    file.write("This line is appended to the file.\n")
    print("Content appended to file.")
```

```python
# Read and Write mode
with open('Sample.txt', 'r+') as file: # Opens the file in read and write mode
    # Reads the content of the file
    content = file.read()
    print("Current content of the file:")
    print(content)
    
    # Writes new content to the file
    file.write("This line is added after reading.\n")
```

## Working with CSV

### CSV Module:
- csv.writer() -> writes data to CSV
- csv.reader() -> Reads CSV rows as a list
- csv.DictReader() -> Reads rows as dictionaries 

```python
# working with CSV 

import csv
# writing to CSV
with open('Sample.csv', 'w', newline='') as csvfile: # Opens file in write mode
    csvwriter = csv.writer(csvfile) # Creates the CSV writer object
    csvwriter.writerow(['Name', 'Age', 'City'])  # Writing header
    csvwriter.writerow(['Alice', 30, 'New York']) # Write data row
    csvwriter.writerow(['Bob', 25, 'Los Angeles'])
    print("CSV file created and data written.")
```

```python
# Reading mode

with open('Sample.csv', 'r' ) as csvfile:
    reader = csv.reader(csvfile)
    for row in reader: # To iterate through each row in the csv file
        print(row)
```

```python
# DictReader for reading csv files

with open('Sample.csv', 'r') as csvfile:
    reader = csv.DictReader(csvfile)
    for row in reader:
        print(row)
```

## Handling JSON Files

- JSON: Format for storing and exchanging date (like Python Dictionaries).
- JSON stores data in Python's dctionary format

### JSON Modules
- json.dump() -> Writes JSON to file
- json.load() -> Reads JSON from file
- json.loads() -> Reads JSON from string
- json.dumps() -> Converts Python object to JSON String

```python
import json

# Write JSON file

data = {
    'name': 'Alice',
    'age': 30,
    'city': 'New York'
}

with open('data.json', 'w') as jsonfile:
    json.dump(data, jsonfile, indent=4) # Writes json data to the file with identation

```

```python
# Reading from JSON File

with open('data.json', 'r') as jsonfile:
    data = json.load(jsonfile)
    print(data)
```

```python
# Reading JSON file from a string

json_string = '{"name": "Alice", "age": 30, "city": "New York"}'
data = json.loads(json_string) # Parse json string into Python Dictionary
print(data)
```

```python
# Converting Python object to JSON String

data = {
    'name': 'Alice',
    'age': 30,
    'city': 'New York'
}

json_string = json.dumps(data, indent=4)
print(json_string)
```
