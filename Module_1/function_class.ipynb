## Function Arguments and Parameters

### Types of Arguments
- Positional Argument: Matched by order
- Keyword Argument: Matched by name (e.g, func(name = "Alice"))

```python
# Positional Arguments and Parameters

def add_positional(a, b, c):
    sum = a + b + c
    return sum

result = add_positional(1, 2, 3)
print(f"The sum of the numbers is: {result}")
```

```python
# Keyword Arguments and Parameters

def add_keyword(name = "Alice", age = 30):
    print("Name:", name)
    print("Age:", age)

add_keyword(name = "Bob", age = 25)  # Using keyword arguments
```

```python
add_keyword(age = 40, name = "charles")
```

### *args Arguments

```python
def add_args(*args):
        print(type(args))  # This will show the type of args as a tuple
        for num in args:
            print(num, end=' ')
```

```python
add_args(1, 2, 3, 4, 5)  # This will print: 1 2 3 4 5
```

### **kwargs Arguments

```python
def add_kwargs(**kwargs):
    print(type(kwargs))  # This will show the type of kwargs as a dictionary
    for key, value in kwargs.items():
        print(f"{key}: {value}")
```

```python
add_kwargs(name="Alice", age=30, city="New York")
```

```python
def average(number):
    total = 0
    for num in number:
        total += num
    avg = total / len(number)
    return avg
```

```python
price = [10, 20, 30, 40, 50]
avg_price = average(price)
print(f"The average price is: {avg_price}")
```

```python
# Get Crypto Price Change

def calculate_price(old_price, new_price):
    if old_price == 0:
        return "Old price cannot be zero."
    change = ((new_price - old_price) / old_price) * 100
    return change 
```

```python
BTC_change = calculate_price(10000, 12000)
print(f"The price changed by {BTC_change:.1f}%")
```

```python
def format_crypto_price(name, price, symbol='$'):
    return f"{name} is currently trading at {symbol}{price:.2f}"

print(format_crypto_price("Bitcoin", 11000))
print(format_crypto_price("Ethereum", 2000, symbol='Ξ'))
```

## Classes

- Classes are blueprints (templates) for creating objects.
- An objects is an instance of a class that can hold data(attributes) and perform actions (methods).
- Classes are ideal for modeling real-world entities like cryptocurrency portfolio

-- Attributes: Variables that store data  
-- Methods: Functions that define the behavior of the class  
-- constructor(init): A special method to initialize objects

--- Use the keyword 'class' to define a class

### Self

- Self is the first parameter in instance methods of a class.  
- It represnts the instance (object) on which the method is called
- Used the access and modify the instance's attribute 

### Defining Classes

```python
class CryptoWallet:
    def __init__(self, owner): # Constructor
        self.owner = owner
        self.balance = {}

    def deposit(self, token, amount):
        self.balance[token] = self.balance.get(token, 0) + amount

    def withdraw(self, token, amount):
        if self.balance.get(token, 0) >= amount:
            self.balance[token] -= amount
            return True
        else:
            return False
        
    def view_balance(self):
        return self.balance
```

```python
wallet = CryptoWallet("Feyi")
wallet.deposit("ETH", 0.7)
wallet.deposit("BTC", 0.1)

print(wallet.view_balance())
```

```python
success = wallet.withdraw("ETH", 0.1)

print("withdrawal success: ", success)
print(wallet.view_balance())
```

### Cryptocurrency and Portfolio Class

```python
class Cryptocurrency:
    """ Class to represent a cryptocurrency """
    def __init__(self, name, price, symbol, quantity):
        self.name = name # e.g Bitcoin
        self.price = price # current price in USD
        self.symbol = symbol # btc
        self.quantity = quantity # amount held

    def get_value(self):
        """ Calculate the total value of the holding """
        return self.price * self.quantity
    
    def updated_price(self, new_price):
        """Updates the cryptocurrency price"""
        self.price = new_price

class portfolio:
    """ Class to manange a portfolio of Cryptocurrency"""
    def __init__(self):
        self.holdings = {} # dictionary to store cryptocurrenncy

    def add_crypto(self, crypto):
        """ Adds a cryptocurrency to a portfolio"""
        self.holdings[crypto.symbol] = crypto

    def get_total_value(self):
        """ Calculates the total value of the portfolio"""
        total = sum(crypto.get_value() for crypto in self.holdings.values())
        return total 
    
    def get_holding(self, symbol):
        """ Retrieves a Cryptocurrency by its Symbol"""
        return self.holdings.get(symbol, None)
    
    def withdraw_crypto(self, symbol, quantity):
        """Withdraw a specifed quantity of a cryptocurrency"""
        crypto = self.get_holding(symbol)

        # Check if cryptocurrency exists
        if not crypto:
            print(f"Error! {symbol} not found in the portfolio")
            return False
        
        # Validate Quantity
        if quantity <= 0:
            print(f"Error! Withdrawal quantity is insufficient.")
            return False
        if quantity > crypto.quantity:
            print(f"Error: Insufficient {symbol} quantity. Available: {crypto.quantity}, Request {quantity}")
            return False
        
        # Update Quantity
        crypto.quantity -= quantity
        print(f"Withdrew {quantity} {symbol}. Remaining {crypto.quantity}")

        # Remove Cryptocurrency if quantity is 0
        if crypto.quantity == 0:
            del self.holdings[symbol]
            print(f"{symbol} holding removed from portfolio")
            return True
```

```python
# Create cryptocurrency object
bitcoin = Cryptocurrency("Bitcoin", 11700, "BTC", 1)
ethereum = Cryptocurrency("Ethereum", 500, "ETH", 0.5)

# Create Portfolio Object
my_portfolio = portfolio()
my_portfolio.add_crypto(bitcoin)
my_portfolio.add_crypto(ethereum)

# Calculate total portfolio values
total_value = my_portfolio.get_total_value()
print(f"Portfolio Total Value: ${total_value:.2f}")
```

```python
# To update the price

bitcoin.updated_price(10000)
print(f"New Bitcoin Price: ${bitcoin.price:.2f}")
print(f"Updated Portfolio Total Value: ${my_portfolio.get_total_value():.2f}")
```

```python
my_portfolio.withdraw_crypto("ETH", 0.2)
print(f"Portfolio Value after withdrawing 0.2 ETH: ${my_portfolio.get_total_value():.2f}")
```
