### Control Structure
Helps you understand the flow of your program. They allow your code to make decisions or repeat actions.

- Loops
- Conditionals

### Conditionals
 A conditional statement controls whether a block of code runs based on whether a condition is true or false
#### Types: 
- If: runs if the condition is true
- elif: Checks for addition if the previous condition is false
- else: Runs if no condition is True

```python
# If Statement Example
x = 10

if x < 20:
    print("x is less than 20")
```

```python
# If-Else Statement Example 
x = 30
if x < 20:
    print("x is less than 20")
else:
    print("x is greater than or equal to 20")
```

```python
# If-Elif-Else Statement Example
x = 7
if x < 5:
    print("x is less than 5")
elif x == 5:
    print("x is equal to 5")
else:
    print("x is greater than 5")
```

```python
Wallet_balance = 1500

if Wallet_balance > 1000:
    print("High Balance")
elif Wallet_balance > 500:
    print("Medium Balance")
elif Wallet_balance == 1500:
    print("Exact Balance")
else:  
    print("Low Balance")
```

### Logical Operators
AND, OR, NOT

```python
v = 100

if v > 0 and v < 200:
    print("v is between 0 and 200")
elif v < 0 or v > 200:
    print("v is either less than 0 or greater than 200")
elif v < 0 and v > 200 or v == 0:
    print("v is either less than 0 or greater than 200, or equal to 0")
else:
    print("v is not in the specified range")
```

### Loops -
### FOR Loop 
Loops are more like repetitions
- FOR Loop: repeats code for each item in the collection
- While loop: repeats code while the condition is true. e.g While hungry, keep eating

```python
## For loop - loop over a list, string, or range

tokens = ['ETH', 'BTC', 'SOL']

for token in tokens:
    print(f"Token: {token}")
```

```python
names = ['Alice', 'Bob', 'Charlie']

for name in names:
    print(name)
```

```python
# type-1 loop

l1 = [44, 55, 66, 77]
total = 0

for i in l1:
    total = total + i
    
print(total)
```

```python
l1 = [44, 55, 66, 77]
total = 0

for i in l1:
    total = total + i
    
Average = total / len(l1)

print('Average is:', Average)
print("Total is:", total)
```

```python
#Type 2- loop
# range() -> returns random numbers

for i in range(2, 11, 2): # prints range of numbers from 2 - 11, with the difference of 2
    print(i)
```

```python
# to get a list of even numbers 
l2 = []

for i in range(20):
    if i%2 == 0:
        l2.append(i)
print(l2)
```

```python
# Type-3 loop - loops through the concept of index

l3 = [22, 33, 44, 55]

for i in range(len(l3)):
    print(l3[i])
```

```python
crypto_prices = {'btc':120000, 'eth':2010, 'sol':125}

# for i in crypto_prices:
#     print(i, crypto_prices.get(i))

# OR

for name, price in crypto_prices.items():
    print(name, price)
```

```python
for i in crypto_prices.keys():
    print(i)
```

### WHILE Loop

```python
# infinity loop -> keeps running while the condition is True

Counter = 0

while Counter < 3:
    print('Counting unless you stop it', Counter)
```

```python
i = 10

while True:
    if i == 20:
        break
    print(i)
    
    i+=1
```

```python
# get even numbers between 1-20

even = []
i = 1

while True:
    if i>20:
        break
    if i%2 == 0:
        even.append(i)
        print(i)
    i += 1
```

```python
# Multiplication Table 3x3

num1 = 3
num2 = 1

while True:
    if num2 == 11:
        break
    print(num1, 'x', num2, '=', num1*num2)
    num2+=1
```

```python
# Continue -> skips to the next iteration of the loop

for i in range(10):
    if i == 5:
        continue
    print(i)
```
