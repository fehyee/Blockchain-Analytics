# Error Handling and Debugging

## Error Handling:

A Python mechanism to manage and respond to errors (exceptions) that occur during program execution, preventing crashes.  

Two ways to handle errors:


### Exception (Try/Except)  
Try: run the code

Except: If it crashes, ignore error or do something else.

Finally: To clean up no matter what happens

```python
# Error Handling using Try/Except

try:
    # Attempt to open a file that may not exist
    with open('file.txt', 'r') as file:
        content = file.read()

except:
    pass  # Ignore the error if the file does not exist

finally:
    # This block will always execute, regardless of whether an exception occurred
    print("This will always run, even if an error occurred.")
```

### Common Errors

- ValueError: when you give the right type of data but it doesn't make sense (Like asking for a negative age)  

- TypeError: making incompatible things (like adding text to a number)  

- FileNotFound: Trying to open a file that doesn't exist  

- KeyError: Looking for a non existent dictionary key
- IndexError

```python
# ValueError Example

number = 'ten'

try: 
    value = int(number)

except ValueError:
    print(f"ValueError: Cannot convert '{number}' to an integer.")
```

```python
# TypeError Example

try:
    result = '5' + 5
except TypeError:
    print("TypeError: Cannot concatenate str and int objects.")
```

```python
# FileNotFound example

try:
    with open('non_existent_file.txt', 'r') as file:
        content = file.read()
except FileNotFoundError:
    print("FileNotFoundError: The file does not exist.")
```

```python
# KeyError Example

my_dict = {'name': 'Alice', 'age': 30}
try:
    value = my_dict['address']
except KeyError:
    print("KeyError: The key 'address' does not exist in the dictionary.")
```

```python
# IndexError Example

my_list = [1, 2, 3]
try:
    value = my_list[5]
except IndexError:
    print("IndexError: List index out of range.")
```
