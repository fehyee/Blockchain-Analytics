## Data Visualization

is the process of representing data graphically to reveal patterns, trends, correlations, and outliers that might be obvious in raw numbers.

### Matplotlib:

A vast Python library used for creating a wide range of plots.  
It gives precise control over every element - Colors, labels, sizes, and more - making it ideal for custom visualization

```python
# Installing/importing matplotlib for plotting graphs

import matplotlib.pyplot as plt
```

```python
# Basic Setup for plotting

students = ['Alice', 'Bob', 'Charlie', 'David']
scores = [85, 90, 78, 92]
```

### Line Plot

```python
plt.plot(students, scores, color = 'blue', marker='o')
plt.title('Student Scores')
plt.xlabel('Students')
plt.ylabel('Scores')
# plt.show()
```

### Scatter Plot

```python
plt.scatter(students, scores, color='red') #, marker='x')
plt.title('Student Scores Scatter Plot')
plt.xlabel('Students')
plt.ylabel('Scores')
plt.show()
```

### Bar Chart

```python
plt.bar(students, scores, color='green')
plt.title('Student Scores Bar Chart')
plt.xlabel('Students')
plt.ylabel('Scores')
plt.show()
```

### Histogram

```python
# plt.hist(scores, color='purple')
plt.figure(figsize=(6, 2)) # To set the size of the plot
plt.hist(scores, bins=5, color='purple', edgecolor='black')
plt.title('Student Scores Histogram')
plt.xlabel('Scores')
plt.ylabel('Frequency')
plt.show()
```

### Figure size

```python
plt.figure(figsize=(6, 2))  # Set the figure size

plt.bar(students, scores, color='green')
plt.title('Student Scores Bar Chart')
plt.xlabel('Students')
plt.ylabel('Scores')
plt.show()
```

### Seaborn

A higher level library built on Matplotlib, designed for statistical data visualization. It simplifies creating complex plots, integrates easily with Pandas DataFrames, and offers attractive default stylesand color palettes.

```python
import seaborn as sns
import pandas as pd
```

```python
# Basic Setup for plotting

students = ['Alice', 'Bob', 'Charlie', 'David']
scores = [85, 90, 78, 92]
```

```python
data = pd.DataFrame({'Students': students, 'Scores': scores})

print(data)
```

### Lineplot

```python
sns.lineplot(x = 'Students', y = 'Scores', data=data, color= 'blue')
plt.title('Students Scores')
plt.show()
```

### Scatter Plot

```python
sns.scatterplot(x = 'Students', y = 'Scores', data=data, color= 'blue')
plt.title('Students Scores')
plt.show()
```

### Bar chart

```python
plt.figure(figsize=(6, 2))  # Set the figure size
sns.barplot(x = 'Students', y = 'Scores', data=data, color= 'green')
plt.title('Students Scores')
plt.show()
```

### Histogram

```python
plt.figure(figsize=(6, 2))  # Set the figure size

sns.histplot(data=data, x='Scores', color= 'purple')
plt.title('Students Scores Histogram')
plt.show()
```

## Visualizing Web3 Token Transfer Data

```python
df = pd.read_csv('web3_token_transfers.csv', parse_dates=['timestamp'])
df['date'] = df['timestamp'].dt.date

# Set Seaborn themes
sns.set(style='whitegrid')
```

```python
df.head()
```

### Token Transfer Volume Distribution

```python
plt.figure(figsize=(7, 3))  # Set the figure size

sns.histplot(df['amount'], bins=50, kde=True, color='skyblue')
plt.title("Distribution of Token Transfer Amount")
plt.xlabel('Amount')
plt.ylabel('Frequency')
plt.xlim(0, 3000)
plt.show()
```

```python
# Daily Total Transfers

daily_volume = df.groupby('date')['amount'].sum().reset_index()

plt.figure(figsize=(10, 4))
sns.lineplot(data=daily_volume, x='date', y='amount', marker='o', color='blue')
plt.title('Daily Total Token Transfer Volume')
plt.xlabel('Date')
plt.ylabel('Total Transfer Amount')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
