## Data Manipulation

The proces of changing, organizing, or transforming data to make it more useful, readable, or suitable for analysis.


## Pandas

- Pandas handles data structure, filtering, sorting or grouping or calculating new values from existing data
- Pandas is used for working with data. Helps load, clean, analyze, and manipulate data easily.

### Key Concepts
- Series: A single column of data (Like a list of labels)
- Labels: Labels from rows, helping you identify and access data

```python
import pandas as pd
```

### DataFrame 

A DataFrame is like a table with rows and columns

```python
# Create a Dictionary with sample data
data = {
    'coin': ['Bitcoin', 'Ethereum', 'Litecoin'],
    'price': [45000, 3000, 150],
    'volume': [1000000, 500000, 200000]
}

df = pd.DataFrame(data)

print(df)
```

### Data Exploration

```python
df.head()  # Display the first few rows of the DataFrame
```

```python
df.tail()  # Display the last few rows of the DataFrame
```

```python
df.info()
```

```python
# Select a specific column

df['coin']  # Access the 'coin' column
```

```python
# Fetching two columns

df[['coin', 'price']]  # Access the 'coin' and 'price' columns
```

```python
# Filtering rows based on a condition

df[df['price'] > 1000]  # Filter rows where price is greater than 1000
```

```python
# Sorting -> sort columns in ascending or descending order

df.sort_values('price') # Sorts in ascending order by default
```

```python
df.sort_values('price', ascending = False) # To sort in descending order
```

```python
# Adding a new column -> Price after 2% change

df['Price_after_2pct'] = df['price'] * 1.02
df
```

```python
df.describe() # Used to investigate
```

```python
# Save and Load data

#Saving to CSV file
df.to_csv('crypto_data.csv', index=False)
```

## Numpy

- Useful for working with arrays (lists of numbers) and performing fast mathematical operations

- Numpy -> Array
- Pandas -> Dataframe

```python
import numpy as np
```

```python
prices = np.array([45000, 50000, 40000, 20000])

print(prices)
```

```python
# To know the number of elements

prices.shape
```

```python
# Data Type check

print(prices.dtype)
```

```python
# adding 1000 to all the prices in the array

prices_plus_100 = prices + 100

print(prices_plus_100)
```

```python
# Computing for 2% increase in price 

price_increase = prices_plus_100 + 1.02
print(price_increase)
```

```python
# Filtering through the array

higher_price = price_increase[price_increase > 30000]
print(higher_price)
```

```python
# Basic statistical functions

volume = np.array([500000, 300000, 700000, 100000])

print("\nStatistics for Volume:")

print("Mean Average: ", np.mean(volume)) # Average volume
print("Minimum: ", np.min(volume)) # Minimum Volume
print("Maximum: ", np.max(volume)) # Maximum Volume
print("Sum: ", np.sum(volume)) # Total Sum of Volume
print("STD (Volatility): ", np.std(volume)) 
```

```python
# Working with a 2D Array

crypto_data = np.array([
    [45000, 50000],
    [33000, 60000],
    [12000, 40000]
])

print(crypto_data)
```

```python
# Access row

crypto_data[1]
```

```python
# Acess column

print(crypto_data[1:,0]) # row 2, column 1
print(crypto_data[:, 1]) # Column 2
```
